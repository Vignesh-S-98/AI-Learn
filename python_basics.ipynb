### Unpacking 
Unpacking lets you take a grouped collection of values and assign them to meaningful variables.

For multiple assignments we can separate the variable names with comma and the values separated by comma. This is called as unpacking 

`x,y,z = 10,20,30` 

We can also pass the values as a tuple or even as a list or set i.e. any collection or Any suitable iterable can be unpacked.

`x,y,z = (10,20,30)` 

the only condition is that the number of values must match the number of variables used. if the no. of variables != no. of values then python raises **ValueError**

unpacking is positional. So the first variable is mapped with the first value and second variable with sec value and so on  

```python
printAttr("Normal")
x, y, z = 10, 20, 30
print(f"x = {x} | y = {y} | z = {z}")

printAttr("Tuple Unpacking")
x,y,z =(10,20,30)
print(f"x = {x} | y = {y} | z = {z}")

printAttr("List Unpacking")
x,y,z = [10,20,30]
print(f"x = {x} | y = {y} | z = {z}")

printAttr("Set Unpacking")
x,y,z = {10,20,30}
print(f"x = {x} | y = {y} | z = {z}")

printAttr("String Unpacking")
x,y,z = "Sai"
print(f"x = {x} | y = {y} | z = {z}")
```

#### Extended Unpacking 

Extended unpacking is an advanced version of unpacking. 

In normal unpacking num(variable) == num(values) else it will raise **Value Error**

But with extended unpacking num(variable) need not be equal to num(values)

We can extract the first value and last value directly using the variable name with the usual unpacking tech, but the middle value can be collected into a single variable using `*` as a prefix for the middle variables. 

We cannot use more than one starred variable for the extended unpacking

The type of the starred variable is always to be a list. 

```python
first, *middle, last = [10, 20, 30, 40, 50]

printAttr("Extended unpacking")
print(f"first value : {first}")
print(f"middle value : {middle}")
print(f"last value : {last}")
```

```python
first, *remaining  = [10, 20, 30, 40, 50]

printAttr("Extended unpacking")
print(f"first value : {first}")
print(f"remaining value : {remaining}")
```

### Object Lifetime 
An object is present in the python memory as long as the object is referenced. Once there is no more reference for the object, it is killed. 

But for integers alone there is a python optimization strategy where the object for the int values -5 to 256 are stored as a cache. 

The reachable means atleast one reference for the object should be present. 

Objects die when the last reference disappears

Python tracks the reference of all the objects by counting the number of references each object is pointed to.

When the count = 0 then the object is removed from the memory. 

But refernce counting and object removal fails when there is circular dependancy is existing.

## Hashing 

Hashing in Python is the process of converting an input of arbitrary size (like a string, number, or tuple) into a fixed-size integer value using a mathematical function. This generated integer, known as a hash value, serves as a unique fingerprint for the data.

Hashing is commonly used in data structures like hash tables, dictionaries, and sets to enable fast data retrieval. 

When you store an object in a hash-based collection, Python computes its hash value and uses it to determine where to store the object in memory.

When you later want to retrieve the object, Python computes the hash value again and looks up the corresponding location.

Hash value is computed using the built-in `hash()` function in Python.

Immutable objects are usually hashable as the values of the objects doesn't change

Mutable objects are usually non hashable.

Dict keys must be hashable. 

Set elements must be hashable  i.e. values inside the sets should be of hashable. 

Hashable means producing a stable hash codes i.e. a code that is same forever / whenever a job is ran.

### Implicit conversion 
Implicit conversion occurs when Python automatically converts one type into another without user casting them manually. 

This examples are 
1. int + float = float -> int is converted to float internally thus it becomes float + float
2. int / int = float 
3. Bool + bool -> int

There is a type hierarchy 

1. bool
2. int
3. float
4. complex

bool gets converted to int, int to float, float to complex based on the internal need

i.e. python converts lower precision to higher precision

Python never converts any type to str or str to any other types implicitly and it raises **TypeError**, when we do any operation btwn the types without conversion

### Explicit Conversion

Explicit conversion happens when user tell Python exactly how to convert a value.

Cases like where the program raises **TypeError** because python cannot perform implicit conversion, we can use the explicit conversion.

When we convert float -> int we lose the decimal values -> this is called *Lossy conversion*. This is usually not advisable unless it what is needed for the logic.

### Logical Operations

| Operator | Meaning                      |
| -------- | ---------------------------- |
| `and`    | Both conditions must be true |
| `or`     | At least one condition true  |
| `not`    | Inverts truth value          |

Mostly we use logical operations for chained comparisons

We can use logical operations to return values 
- logical AND returns the first falsy value or the last value if all are truthy.
- logical OR returns the first truthy value or the last value if all are falsy.


Operator precedence: not > and > or ->  python evaluates not first and then and then or

#### Short circuit evaluation 

Python can determine the result of a logical expression before evaluating every part.

For `and` if first value is false then the evalution of the second value is skipped 

For `or` if the first is true then the evalution of the second value is skipped 

This is the reason why when we use 
- `0 and 100` we get `0` 
- `True or print("Hello")` we get `True`

This is handy for some cases of `None` check, when we have any variable value as `none` and we try to apply any function on it, it will raise **AttributeError**, so we can write as `variable and variable.attribute` now since none is falsy the evaluation wont happen and the error is not raised

The order of short-circuiting i from *left to right*

```python
printAttr("Logical Operations")
print(f"Logical AND: {True and False}")
print(f"Logical OR: {True or False}")
print(f"Logical NOT: {not True}")
printAttr("Logical operators returning values",'*')
print(f"Logical AND: {10 and 20}")
print(f"Logical OR: {10 or 20}")
print(f"Logical NOT: {not 10}")
printAttr("Operator precedence",'*')
print(f"not False and True: {not False and True}") 
printAttr("Short circuiting",'*')
print(f"Logical AND with 0: {10 and 0}") 
print(f"Logical OR with 0: {10 or 0}") 
print(True or print("This will not be printed"))
```
