# 📚 DataFrame Interview Notes

---

# 1. DataFrame Basics

## Q1. What is a DataFrame?

**Answer**

A DataFrame is a **two-dimensional labeled data structure** in Pandas consisting of:

- Row Index
- Column Index
- Data Values

It is similar to a spreadsheet or SQL table but supports **heterogeneous column types**.

---

## Q2. Can a DataFrame contain different data types?

**Answer**

Yes.

Each column can have its own dtype.

Example:

| Column | dtype |
|---------|-------|
| Name | string |
| Age | int64 |
| Salary | float64 |
| Joined | datetime64 |

---

## Q3. Why is Pandas considered column-oriented?

**Answer**

Because operations are optimized around columns. Values in a column are stored together, allowing efficient vectorized computations using NumPy.

---

## Q4. Why are vectorized operations faster than row-wise loops?

**Answer**

Vectorized operations execute on entire arrays/columns using optimized NumPy routines implemented in C, avoiding the overhead of Python loops and making them significantly faster.

---

# 2. Creating DataFrames (`pd.DataFrame()`)

## Q1. What is the purpose of `pd.DataFrame()`?

**Answer**

It constructs a DataFrame from various data sources such as:

- Dictionaries
- Lists
- NumPy arrays
- Series
- Structured arrays
- Another DataFrame

---

## Q2. What are the parameters of `pd.DataFrame()`?

**Answer**

```python
pd.DataFrame(
    data=None,
    index=None,
    columns=None,
    dtype=None,
    copy=None
)
```

---

## Q3. Which parameter is the most important?

**Answer**

`data`

Without it, there is usually nothing meaningful to construct.

---

## Q4. Why can `pd.DataFrame()` accept many input types?

**Answer**

Because the constructor first **normalizes** different input types into Pandas' common internal tabular representation before constructing the DataFrame.

---

## Q5. What happens internally when you call `pd.DataFrame()`?

**Answer**

```text
Receive Input
      ↓
Identify Input Type
      ↓
Normalize Input
      ↓
Determine Shape
      ↓
Validate Labels
      ↓
Infer / Apply dtypes
      ↓
Allocate Memory
      ↓
Return DataFrame
```

---

## Q6. How does Pandas determine the shape of a DataFrame?

**Answer**

The shape is determined primarily from the **data**.

After determining the shape, Pandas validates the supplied **index** and **columns**.

---

## Q7. Does Pandas internally store dictionaries?

**Answer**

No.

The dictionary is only an input format.

Internally, Pandas converts it into its own optimized column-based data structures.

---

# 3. Constructor Parameters

## Q1. What happens if you don't specify an index?

**Answer**

Pandas creates a default `RangeIndex`.

Example:

```text
0
1
2
3
...
```

---

## Q2. Why must the index length match the number of rows?

**Answer**

Because every row needs exactly one row label.

Otherwise Pandas cannot correctly map labels to rows.

---

## Q3. What happens if you don't specify columns?

**Answer**

It depends on the input.

- Dictionary → keys become column names.
- List/NumPy array → numeric column labels are generated.

---

## Q4. Why must the number of column labels match the number of columns?

**Answer**

Each column needs exactly one label.

If the counts don't match, Pandas raises a **ValueError**.

---

## Q5. How does `columns` behave differently for dictionaries and lists?

**Answer**

For **lists/arrays**:

- Assigns column names.

For **dictionary-like inputs**:

- Selects columns.
- Reorders columns.
- Missing requested columns are created and filled with `NaN`.

---

## Q6. What happens if `dtype=float` is given but one column contains strings?

**Answer**

Pandas attempts to convert every column to `float`.

If conversion is impossible, the constructor raises an exception.

---

## Q7. Why is `copy=False` only a preference?

**Answer**

Memory can only be shared when the input layout is compatible with Pandas' internal representation.

If sharing isn't possible, Pandas creates a copy anyway.

---

# 4. Memory & Performance

## Q1. Why is a NumPy array usually more memory-efficient than a list of lists?

**Answer**

NumPy arrays already use contiguous, homogeneous memory layouts that align well with Pandas' internal storage.

Lists of lists are collections of Python objects and often require additional normalization and copying.

---

## Q2. Why should unnecessary copying be avoided?

**Answer**

Because copying increases:

- Memory usage
- CPU time
- Execution time

This becomes especially important with large datasets.

---

## Q3. When would you explicitly specify `dtype`?

**Answer**

When you need consistent storage types, such as:

- Machine learning
- Database export
- Schema consistency
- Data validation

---

# 5. Scenario-Based Questions

## Q1. You receive JSON from an API. Which constructor input is most natural?

**Answer**

A **dictionary** (or **list of dictionaries** for multiple records), because JSON objects naturally map to Python dictionaries.

---

## Q2. You already have a NumPy feature matrix. How would you create a DataFrame?

**Answer**

Pass the NumPy array as `data` and provide meaningful column names using the `columns` parameter.

---

## Q3. You need employee IDs as row labels. Which parameter do you use?

**Answer**

Use the `index` parameter.

Example:

```python
pd.DataFrame(
    data,
    index=["EMP101", "EMP102"]
)
```

---

## Q4. You only want three columns from a dictionary containing twenty columns.

**Answer**

Use the `columns` parameter to select only the required columns during construction.

---

## Q5. You want every numeric value stored as `float64`. Which parameter should you use?

**Answer**

Use:

```python
dtype=float
```

provided all columns can be converted successfully.

---

# 6. Production & Debugging

## Q1. Why is understanding constructor behavior important in ETL pipelines?

**Answer**

Because data ingestion is the first stage of an ETL process.

Correct DataFrame construction ensures:

- Correct schema
- Proper labels
- Appropriate data types
- Efficient memory usage

before downstream transformations.

---

## Q2. Why is understanding the constructor useful for debugging?

**Answer**

Many issues originate during DataFrame construction, such as:

- Unexpected `NaN` values
- Incorrect labels
- Mixed data types
- Shape mismatches

Understanding how the constructor interprets its parameters makes these problems much easier to diagnose.

# DataFrame Interview Questions 1

## Data Inspection

### Q1. What is the difference between `head()` and `tail()`?

**Answer**

- `head()` returns the first *n* rows.
- `tail()` returns the last *n* rows.
- Both return a new DataFrame.
- Neither modifies the original DataFrame.

---

### Q2. Why would you use `sample()` instead of `head()`?

**Answer**

`head()` always shows the beginning of the dataset, whereas `sample()` returns random rows. `sample()` is useful for exploratory data analysis because issues may exist anywhere in the dataset.

---

### Q3. What information does `info()` provide?

**Answer**

`info()` displays:

- Number of rows
- Number of columns
- Column names
- Data types
- Non-null counts
- Memory usage

---

### Q4. What is the difference between `info()` and `describe()`?

**Answer**

- `info()` provides the structural summary of the DataFrame.
- `describe()` provides statistical summaries of the data.

---

### Q5. Does `describe()` include string columns by default?

**Answer**

No.

By default, `describe()` summarizes only numeric columns. String/object columns can be included using the `include` parameter.

---

### Q6. What is the purpose of `memory_usage()`?

**Answer**

It reports the memory consumed by each column (and optionally the index), helping identify opportunities for memory optimization.

---

### Q7. What is the difference between `unique()` and `nunique()`?

**Answer**

- `unique()` returns the distinct values.
- `nunique()` returns the count of distinct values.

---

### Q8. What is the difference between `nunique()` and `value_counts()`?

**Answer**

- `nunique()` counts how many distinct values exist.
- `value_counts()` counts how frequently each distinct value occurs.

---

### Q9. How does `DataFrame.value_counts()` differ from `Series.value_counts()`?

**Answer**

- `Series.value_counts()` counts occurrences of values in a single column.
- `DataFrame.value_counts()` counts occurrences of complete row combinations.

---

### Q10. What does `duplicated()` return?

**Answer**

A Boolean Series where:

- `False` indicates the first occurrence or a unique row.
- `True` indicates a duplicate row.

---

### Q11. Does `duplicated()` remove duplicate rows?

**Answer**

No.

It only identifies duplicate rows. To remove duplicates, use `drop_duplicates()`.

---

### Q12. Which methods are commonly executed immediately after reading a CSV file?

**Answer**

A common inspection workflow is:

- `head()`
- `info()`
- `describe()`

These methods help verify the imported data before further analysis.
