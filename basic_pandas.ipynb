# Renaming 
## Instance column rename
`df.rename(columns={"actual name" : "new name"})`

Note:- this doesn't update the dataframe but only the instance. To save the updated instance, we can refer that to a variable.

```python
renamed_df = df.rename(columns={
    "first_name" : "First Name",
    "last_name" : "Last Name",
    "phone_number" : "phone number".title(),
    "joining_date" : "joining date".title()
})
renamed_df.head()
```

```python
df.rename(columns={
    "first_name" : "First Name",
    "last_name" : "Last Name",
    "phone_number" : "phone number".title(),
    "joining_date" : "joining date".title()
}).head()
```

```python
df.head()
```

## Permanent column rename
`df.rename(columns={"actual name" : "new name"},inplace=True)`

Note:- this update the dataframe and store in the same.

```python
df2 = df.copy(deep=True)
df2.head()
```

```python
df2.rename(columns={
    "first_name" : "First Name",
    "last_name" : "Last Name",
    "phone_number" : "phone number".title(),
    "joining_date" : "joining date".title()
},inplace=True)
df2.head()
```

## Shallow copy 
Modifying the structure (adding/deleting columns or changing cell values via .loc) in df2 will not affect df.    
But by default in standard Python, a shallow copy can sometimes still mutate nested Python objects (like lists or dicts stored inside cells) in the original DataFrame.    
This is the standard way to safely duplicate a DataFrame for normal data analysis steps without worrying about the SettingWithCopyWarning

`df2 = df.copy()`



```python
og_df = pd.read_csv(r'techcorp_workforce.csv')
og_df.head()
```

```python
copy_df2 = og_df.copy()
copy_df.head()
```

```python
copy_df.rename(columns={
    "first_name":"first name".title()
},
inplace=True)
copy_df.head()
```

# Dupliactes
## Remove Duplicates    
Removes the rows with the duplicate values for the given column

`df[['column_name']].drop_duplicates()` 

This works on a dataframe and return a dataframe

```python
orders_df = pd.read_csv(r'orders.csv')
orders_df.head() 
```

```python
odf1= orders_df.copy(deep=True)
unique_details_df = odf1[["order_details"]].drop_duplicates()
type(unique_details_df)
```

## Multiple column names 
When multiple columns passed, it removes rows where the entire combination is duplicated:

```python
odf1[['order_details','cust_id']].drop_duplicates().head()
```

When a single column needs to be worked on we can use
