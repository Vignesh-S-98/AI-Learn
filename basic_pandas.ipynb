# Filtering Rows 
Get the rows that are having values satisfying a certain criteria.

This is called as boolean indexing.

`df[df["column_name"] condition "criteria"]`


The expression inside the bracket produces a Series of True/ False values 

Conditions can be `== <= < >= > !=` 

```python
df[df["department"] == "HR"]
```

### Internal Working

`df[df["department"] == "HR"]` <- produces a `series` of `True/ False `

This `Series` of `True` is compared with each row of the `dataframe` <- `df[boolean_series]`

The true rows are displayed. 

```python
df[df[["department"]] == "HR"]
```

```python
df[["department"]] == "HR"
```

### Internal Working

`df[df[["department"]] == "HR"]` <- produces a `dataframe` of `True/ False `

This `dataframe` creates a masking effect on the original dataframe where   
`if mask[i, j] == True:`     
`    keep value`        
`else:`          
`    set NaN`         

so pandas aligns by 
1. row index 
2. column names 

when column name matches between the 2 df (the masking and the original df) then that column gets the value. 

When the column name doesn't match `nan` is filled

```python
odf1[df[["department"]]=="HR"]
```

# Multiple conditions

Instead of using just a single boolean filter, we can use multiple filters to filter the rows.

we use `& , |, ~` instead of `and, or, not`

This is because the former is an elementwise operator while the later is the scalar operator. 

elementwise operator compares each rows while scalar  operator compares the whole table. 

For filtering we need to compare each rows rather than the whole table. 

## And conditions

```python
odf1[
    (odf1["order_details"] == "Coat") &
    (odf1["total_order_cost"] > 100)
]
```

## Or conditions

```python
df[
    (df.department == "HR") |
    (df.department == "Admin")
    ]
```

## Or on same column 
We can use `.isin()` for multiple or conditions

```python
df[df["department"].isin(["HR", "Admin"])]
```

# Full Pattern 
In practice, you almost never sort alone. 

The real pattern is a chain: 
1. filter to the rows you care about, 
2. sort by the column that matters, 
3. limit to the top N, and 
4. select only the columns you need. 

This chain is the backbone of exploratory analysis.

```python
df[df["department"] == "Engineering"].sort_values("joining_date", ascending=False).head(5)[["first_name", "last_name", "joining_date"]]
```

## Multiple functions at once 

to use `sum, max, min, mean, count, etc`  we can use `.agg([ _list_of_functions_ ])`

`.agg()` is a function that takes a **Series** and returns a **scalar OR Series**

```python
odf1["total_order_cost"].agg(["sum", "mean", "min", "max","count"])
```

```python
df.columns
```

```python
df['salary'].agg(['min', 'mean', 'max']).rename(
    index={'min': 'lowest_salary', 'mean': 'mean_salary', 'max': 'highest_salary'}
).to_frame().T
```

```python
sales_df[sales_df['salesperson'].str.lower().isin(["samantha","lisa"])]['sales_revenue'].sum()
```

```python
df[df['salary']==df['salary'].max()]['first_name']
```

We can add custom function inside `.agg()` function 

```python
print(df['salary'].agg(['mean', 'max', lambda x: x.min()]))
print("Normal:\n",df['salary'].agg(['mean', 'max', 'min']))
print("Custom:\n",df['salary'].agg(['mean', 'max', lambda x: x.max() - x.min()]))
```
