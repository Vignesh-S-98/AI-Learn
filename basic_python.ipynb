# List Comprehensions
## Creating a list from another list 

`[ _value_ for _value_ in _iterable_ if _value passes a condition_ ]`

```python
num = [1,2,3,4,5,6]
odd =[]
# get odd numbers 
for n in num :
    if n%2 != 0:
        odd.append(n)
print(odd)

#using list comprehension
odd = [n  for n in num if n % 2 != 0]
print(odd)
```

## Creating tuple of list from another list 

with if else condition we need to bring the condition front before the loop

```python
odd__or_even = [(n,"Odd")  if n % 2 != 0 else (n,"Even") for n in num]
print(odd__or_even)
```

## Looping with a function

There are 2 ways thorugh wich we can loop through the function 
1. get true or false <- `.filter()`
2. apply the function <- `.map()`     

```python
def is_odd(n):
    return n%2 ==0
odd = filter(is_odd,num)
print(odd)
print(list(odd))
```

```python
def square(n):
    return n**2
square_nums= map(square,num)
print(square_nums)
print(list(square_nums))
```

## Sum of list elements 
1. add just the numbers inside the list <- `sum(iterable)`
2. add the numbers but not start from 0 <-  `sum(iterable,start)`

```python
print(sum(num))
print(sum(num,100))
```

# Lambda Functions

Lamda functions are one line functions 

Use `lambda variable : variable_function`

```python
odd = list(filter(lambda n : n%2 !=0, num))
print(odd)
```
