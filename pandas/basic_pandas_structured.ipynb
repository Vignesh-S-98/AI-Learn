# Section 2:-  Series

Series is an 1-dimensional labeled array capable of holding any data type.

A Series is backed by an array-like structure. Arrays are homogeneous.

A Series has a single dtype. If all values can be represented by one efficient dtype (like `int64` or `float64`), Pandas uses that. If not, it uses `object` (or another suitable extension dtype) so every element still has the same dtype.

Series has 3 important components

1. Values → actual data (data)
2. Index → labels for each data point (metadata)
3. Dtype → datype of the values (metadata)


index by default is auto generated but can be customized 

The auto-generated index is a *RangeIndex* starting from 0 to n-1 where n is the number of values in the series

RangeIndex or even the customized index are metadata that is used to access the values in the series, they are not the part of the actual data. 

both values and index can contain any datatypes 

index and value need not be homogenous 

`len(index) == len(value)`

So unlike for list, dict, tuple, pandas doesn't want to look for each element to determine the dtype, numpy is already int/ float

The real difference between numpy and series is 

| Feature               | NumPy Array                                                         | Pandas Series                                                                                      |
| --------------------- | ------------------------------------------------------------------- | -------------------------------------------------------------------------------------------------- |
| **Data type (dtype)** | One dtype for the entire array                                      | One dtype for the entire Series (except `object` or extension dtypes can hold more general values) |
| **Index**             | Positional index (0, 1, 2, ...)                                     | Has an explicit index (can be labels like `"A"`, `"B"`, etc.)                                      |
| **Arithmetic**        | Position-based                                                      | **Index-aligned**                                                                                  |
| **Missing values**    | Limited (`NaN` only for floats, newer nullable types in some cases) | Better support (`NaN`, `<NA>`, nullable dtypes)                                                    |
| **Purpose**           | Fast numerical computing                                            | Labeled data analysis                                                                              |

### 2.3.5 Index Alignment

The important need for the index in series is to maintain the alignment. 

When we use numpy all the operations were based on the position and not the index 

So pandas series uses the index to maintain the proper corresponding values are involved in the computation

Alignment is Pandas' automatic process of matching data based on index labels before performing operations. 

Instead of combining values by their positions, Pandas first aligns records with the same index labels, ensuring that calculations are performed on logically corresponding data. 

If a label is missing in one Series, the result for that label becomes NaN rather than producing an incorrect calculation.

````python
# without the custom index - order not maintained

salary = pd.Series(
    [50000,60000,70000])

bonus = pd.Series(
    [0.7,0.5,0.6])

print(salary+bonus)
````

````python
# with custom index - order maintained

salary = pd.Series(
    [50000,60000,70000],
    index=["Alice","Bob","Charlie"]
)

bonus = pd.Series(
    [0.7,0.5,0.6],
    index=["Charlie","Alice","Bob"]
)

print(salary+bonus)
````

When the series are of different size, now the missing index gets`Nan` value as one series is not able to find that index

in positional based, the last value gets `Nan`

````python
salary = pd.Series(
    [50000,60000,70000])

bonus = pd.Series(
    [5000,7000])
print(salary+bonus)
````

````python
salary = pd.Series(
    [50000,60000,70000],
    index=["Alice","Bob","Charlie"]
)

bonus = pd.Series(
    [5000,7000],
    index=["Alice","Charlie"]
)
print(salary+bonus)
````

## 2.4 Attributes in Series 

Attributes are properties of an object that provide information about its state or characteristics. 

For a Pandas Series, attributes often give metadata about the Series.

These attributes, based on the meta data it provides we can classify them as 

1. Structural
2. Metadata 
3. Advanced  

These are based on the data they return and is not structured likely in python. Advanced ttributes are rarely used in the day-to-day activity 


1. Structural Attributes 

    a. `.index` <- returns type *Index or Rangeindex* with index of the Series 

    b. `.value` <- returns type *numpy.ndarray* the underlying data of the Series as a NumPy array.

    It is not advisable to do operations on using the values alone as, 
    Using values removes the index (identity layer), so Pandas can no longer perform automatic alignment. Subsequent operations are performed positionally like NumPy arrays, which can lead to incorrect results if the data is not in the same order.

    c. `.dtype` <-  Describes the dataype of the entire series object. 

    d. `.shape` <- pandas always returns a *tuple* that contains the dimension of the object i.e. in series (no. elements,) in the series 

    To create a tuple with one element, Python requires a trailing comma. Hence the trailing comma.

    This changes for dataframes as (no. rows, no. columns)

    Shape thus describes how the elements are arranged

    e. `.size` <- return the *int* of how many individual values exist i.e. total no. of elements 

    f. `.ndim` <- returns no. of dimensions, this is 0 for a scalar, 1 for series and 2 for dataframe  


2. Metadata Attributes 

    a. `.name` <- the constructor takes `name` arg, this attribute returns that name of the series 

    b. `.empty` <- returns a boolean response on whether the series has empty values (no values) or has any values inside it. 

    If all the values are `Nan` in the series then also it is **not considered as an empty series**.

    Empty meanns no rows exist at all. So there should be absolutely no values while `nan` is considered to be a value.


    c. `.hasnans` <- returns *boolean* response if there are any presence of `Nan` in the series. 

    Even a single `nan` in the list will provide this to be `True`


3. Advanced Attributes

    a. `.array` <- returns the *pandas* extension array of values of a series.

    Even thou `.values` returnt he array of values they are represented by numpy array, but pandas has their own array structure. 

    Numpy cannot always be represented perfectly for thefollowing imporovements in pandas 
    - Nullable integers (Int64)
    - Nullable booleans
    - StringDtype
    - Categorical data
    - Arrow-backed arrays

    thus we use pandas extension array 

    b. `.flags` <- returns an object containing internal implementation flags.

    These flags help Pandas manage certain behaviors and internal guarantees.

    this is almost never used.

`NaN`mean not a number, this means it is a missing value. It is a placeholder for missing information, not an actual number.

We use this from numpy library as pandas is originally built over numpy 

To create a numpy `nan` in pandas or in numpy array, we can use `np.nan`

Nan is a special floating-point value understood by NumPy and Pandas.

`Nan` is not same as `None`, None is *no object*, yet pandas can convert `none` into `nan`

`None` is a Python object. `NaN` is a special numeric missing value.

The need for conversion of `None` to `nan` inside pandas is when we try to perform any mathematical operation with `None`, we get **TypeError** and the calculation crashed. 

The reason for the datatype changing from int to float is that `nan` is basically from *IEE Floating numbers standard*.  A Series needs to be in a homogenous datatype. Hence the entire series is converted into a floating  datatye rather staying as int. 

It can keep it as an object, but it chose to keep float because object calculation is time consuming, Python has to inspect every object individually.

In cases like these, when we have int type but due to null values it is as float, the recent pandas has `pd.Nan` and not `np.Nan`. This keeps the column to be as int. 

This is not a default one but is done by using the arg `dtype="Int64"` inside the series constructor. 

Here `Int` is the class name for `int` type and `64` is the bites, we can use `8`, `16` and `32` based on the data.

The reason is for ata to scientists choose the uppercase Int64 format is for memory savings, precise matching, and data integrity.

Data processing is rarely just about raw math speed; it is also about memory limits and data accuracy.

### 2.8.4 Handling `Nan`

There are 3 common strategies to handle the missing values 

1. Remove the record  <- `dropna()`
2. Replace with a value  <- `fillna()`
3. fill it with neighbouring values <- (forward fill or backward fill)

#### 2.8..4.1 Remove the reocrd
We drop the `nan` records from the series by calling `.dropna()` on the series 

this will not modify the existing series but returns a *new series* 

for positional indexes if we perform `dropna()` the positional value doesn't change, the reason being that they are positional labels and not row numbers, so the label is not changed or updated automatically. 

To reset the postiotional index to be in theproper order again we need to perform `.reset_index(drop = True)`, the `drop = True` arg is important as the older index will become the separate column thus becoming a dataframe

If we have no `nan` in the series there is no error returned

If we have all the values as `nan` then the empty series is returned. 

````python
# series 
s = pd.Series([100, np.nan, 300])
print(s)
print("-"*10)

# droping 
s.dropna()
print(s)
print("-"*10)

#droping and assigning
s = s.dropna()
print(s)
print("-"*10)

#reseting the index
s.reset_index()
print(s)
print("-"*10)

# reseting and assigning
a=s.reset_index()
print(a)
print(type(a))
print("-"*10)

# reset and assign with drop
b=s.reset_index(drop=True)
print(b)
````

#### 2.8..4.2 Replace with Fixed value 

When we replace the values of `nan` with a fixed or calculated value we can use `.fillna()` on the series

Return a *new series* and will not affect the original series 

Since no data is lost, there is no change in the index/ label

There are 2 ways of filling 
1. use a fixed value <- but this is not suitable for all the scenarios eg. `.fillna(0)`
2. use a statistical value 

    There are 3 common statistical methods to fill the `nan`

    - `.fillna(s.mean())` <- used when the data is continuous and numeric / distribution is symmetric / no major outliers
    - `.fillna(s.median())` <- only on numeric data / skewed distribuion / outliers is there
    - `.fillna(s.mode()[0])` <- used when the data is categorical / outliers is there 

    For mode <- there can be multiple frequently occuring data with the same frequency. So we use this separate calculation of mode on a series and use the first output.

    If we didnt provide [0] for mode then the last value so mode returns a series and the last value of the series is considered to be the value to replace 



````python
s = pd.Series([100, np.nan, 300, 200, 100,200])
print(s)
print("-"*10)


# filling 
print("filling".upper())
s.fillna(0)
print(s)
print("-"*10)

# filling and assigning
print("filling and assigning fixed value ".upper())
a = s.fillna(0)
print(a)
print("-"*10)


# filling and assigning
print("filling Mean".upper(),s.mean())
b = s.fillna(s.mean())
print(b)
print("-"*10)


# filling and assigning
print("filling median".upper(),s.median())
c = s.fillna(s.median())
print(c)
print("-"*10)


# filling and assigning
print("filling mode".upper(),s.mode())
d = s.fillna(s.mode())
print(d)
print("-"*10)


# filling and assigning
print("filling mode[0]".upper(),s.mode()[0])
e = s.fillna(s.mode()[0])
print(e)
print("-"*10)
````

#### 2.8..4.3 Replace with previous or next values

When a series has values where there is no sudden jump between the 2 values we can use this approach 

1. If a `nan` value needs to be fixed based on the previous value, then we use `.ffill()` 

   This will fill all the `nan` values with the previous value 

   If the first value is also `nan` then it cannot fill that 


2. If a `nan` value needs to be fixed based on the nect value, then we use `.bfill()` 

   This will fill all the `nan` values with the next value 

   If the last value is also `nan` then it cannot fill that




````python
s = pd.Series([np.nan,2,np.nan,np.nan,3,np.nan,7,np.nan])
print(s)
print("-"*10)


# filling with forward fill
print("forward filling".upper())
a = s.ffill()
print(a)
print("-"*10)

# filling with backward fill
print("backward filling".upper())
b = s.bfill()
print(b)
print("-"*10)
````

### 2.9.1 Aggregation Methods

Used to summarize the data by producing a single result, such as the sum, average, minimum, maximum, or count of values

To have a better model we can classify these aggregation methods into mini-groups 
1. Counting <- How many values do I have
2. Arithmatic <- Whats the total/average
3. Extremes <- What's the smallest/largest
4. Dispersion <- How spread out is the data

**Counting:-** 

1. `.count()` <- return *int* of number of **non-null values** in the series

    The common differnece between `count()` and `size` is that `size` includes non-null values while `count()` excludes them 
    
**Arithmatic:-**

2. `.sum()` <- returns a *float* value of the total (addition) of all the values in the series. This excludes `nan` so that, even if there is `nan` value in the series the result is an integer and not `nan`. 

    Sum of a series with only `nan` is also 0 and not `nan`

3. `.mean()` <- returns a *float* value of the average of series. This internally is equivalent of sum()/count(), hence this also exculdes the `nan` values.

    But when the series is having only `nan` values then sum =0  and count =0 hence the mean will be `nan`


**Extremes:-**

4. `.max()` <- returns a *float* the largest value in the series

5. `.min()` <- returns a *float* the smallest value in the series

    in both min and max `nan` is excluded. 

    For as series with only `nan`, then `nan` is returned 

    Pandas does not need to sort the entire Series. It can determine the smallest/ largest value by scanning once, making it more efficient than sorting just to find an extreme.

**Dispersion:-** This provides how far each value is from the mean, thus we get a better understanding of the data spread

    variance is the average squared distance of each values from the mean 

7. `.std()` <- return a *float* of standard-deviation i.e. sqroot of variance. 

    standard-deviation is the average actual distance from the mean

Statistically low dispersion means better or consistent data. 

Meaning :- 
    If variance or standard-deviation is closer to the mean then the dispersion is low.

# Section 3 :- Interview for Series

## 3.1 — Why Series?

### **Q1. Why isn't a Python list sufficient for data analysis?**

**Answer**

- Lists don't support efficient vectorized operations.
- Lists don't have meaningful labels (only positions).
- Data alignment isn't possible.
- Numerical computations are slower compared to NumPy-backed Pandas objects.

---

### **Q2. Why use a Series instead of a list?**

**Answer**

A Series combines:

- Values
- Index (labels)

This allows:

- Label-based access
- Automatic alignment
- Vectorized operations
- Missing value handling.

---

## 3.2 — Anatomy of a Series

### **Q3. What are the three components of a Series?**

**Answer**

- Values
- Index
- `dtype`

---

### **Q4. What is metadata?**

**Answer**

Metadata is **data that describes the actual data**.

Examples:

- Index
- `dtype`
- Shape
- Size

---

### **Q5. What happens if no index is provided?**

**Answer**

Pandas automatically creates a **RangeIndex** starting from **0**.

---

## 3.3 — Index

### **Q6. Why isn't the index just row numbers?**

**Answer**

The index provides the **identity** of the data.

- Positions may change.
- Labels represent what the data belongs to.

---

### **Q7. What is RangeIndex?**

**Answer**

The default integer index automatically created by Pandas.

---

### **Q8. Why use a custom index?**

**Answer**

To represent meaningful labels instead of positions.

Examples:

- Employee IDs
- Student Roll Numbers
- Product Codes

---

### **Q9. Can a Series have duplicate indices?**

**Answer**

Yes.

Pandas allows duplicate indices.

However, certain operations become ambiguous.

---

### **Q10. Why is an Index immutable?**

**Answer**

Because it provides:

- Protection against accidental modification.
- Safe sharing between multiple Series/DataFrames.
- Better performance.
- Reliable alignment.

---

### **Q11. What is Index Alignment?**

**Answer**

During arithmetic operations, Pandas matches values using **labels** rather than positions.

---

### **Q12. When should we reset an index?**

**Answer**

Reset the index after:

- Filtering
- Dropping rows
- Sorting

when sequential numbering is desired.

---

## 3.4 — Series Attributes

### **Q13. What does `.index` return?**

**Answer**

Returns the **Index object**.

---

### **Q14. What does `.values` return?**

**Answer**

Returns the underlying **NumPy array**.

---

### **Q15. Why use NumPy internally?**

**Answer**

Because NumPy provides:

- Fast computations
- Efficient memory usage
- Vectorized operations

---

### **Q16. What does `.dtype` return?**

**Answer**

Returns the datatype of the stored values.

---

### **Q17. Difference between `.shape`, `.size`, and `.ndim`**

**Answer**

| Attribute | Meaning |
|-----------|---------|
| `.shape` | Dimensions |
| `.size` | Number of elements |
| `.ndim` | Number of axes |

---

## 3.5 — Accessing Data

### **Q18. Difference between `.loc` and `.iloc`**

**Answer**

| `.loc` | `.iloc` |
|--------|---------|
| Label-based | Position-based |
| Uses index labels | Uses integer positions |

---

### **Q19. Difference between `.at` and `.iat`**

**Answer**

Both access a **single value**.

| `.at` | `.iat` |
|-------|--------|
| Label-based | Position-based |

They are optimized for **scalar access**.

---

### **Q20. What is Boolean Indexing?**

**Answer**

Filtering using a Boolean Series.

**Example**

```python
s[s > 100]
```

---

## 3.6 — Updating Data

### **Q21. Can you create new labels while assigning?**

**Answer**

Yes.

```python
s["C"] = 100
```

If the label doesn't exist, Pandas automatically creates it.

---

## 3.7 — Vectorization

### **Q22. What are vectorized operations?**

**Answer**

Operations performed on the **entire Series at once** without explicit loops.

---

### **Q23. Why are vectorized operations faster?**

**Answer**

Because Pandas delegates computation to optimized NumPy implementations instead of Python-level loops.

---

## 3.8 — Automatic Alignment

### **Q24. Does Pandas align by position?**

**Answer**

No.

Pandas aligns data using **index labels**, not positions.

---

### **Q25. What happens when labels don't match?**

**Answer**

- The union of labels is formed.
- Missing matches become `NaN`.

---

## 3.9 — Missing Values

### **Q26. What is NaN?**

**Answer**

A special floating-point value used by Pandas to represent **missing or unknown data**.

---

### **Q27. Why not use 0 instead of NaN?**

**Answer**

Because **0 is a valid value**, whereas **NaN specifically indicates missing information**.

---

### **Q28. Why does an integer Series become `float64` after introducing NaN?**

**Answer**

A Series stores one homogeneous datatype.

Traditional `int64` cannot represent `NaN`, so Pandas performs **type promotion (upcasting)** to `float64`.

---

### **Q29. Is `"NaN"` the same as `np.nan`?**

**Answer**

No.

- `"NaN"` → String
- `np.nan` → Special floating-point value

---

### **Q30. Is `None` the same as NaN?**

**Answer**

No.

- `None` is a Python object.
- `NaN` is a floating-point representation for missing values.

Pandas often converts `None` to `NaN` in numeric Series.

---

### **Q31. Why is `np.nan == np.nan` False?**

**Answer**

Because `NaN` represents an **unknown value**.

Two unknown values are **not guaranteed to be equal**.

---

### **Q32. Why shouldn't we use `== np.nan`?**

**Answer**

Because comparisons with `NaN` always return `False`.

Use:

```python
isna()
```

instead.

---

### **Q33. Difference between `isna()` and `notna()`**

**Answer**

| Function | Purpose |
|----------|---------|
| `isna()` | Detect missing values |
| `notna()` | Detect non-missing values |

---

### **Q34. What does `s[s.isna()]` do?**

**Answer**

Returns only the rows whose values are missing.

---

### **Q35. How does NaN behave in arithmetic?**

**Answer**

Arithmetic involving `NaN` generally propagates `NaN`.

Example:

```python
NaN + 5   # NaN
NaN * 10  # NaN
```

---

### **Q36. How do aggregate functions treat NaN?**

**Answer**

Most aggregate functions ignore `NaN` by default, including:

- `sum()`
- `mean()`
- `min()`
- `max()`

**Special cases**

- `sum()` of an all-NaN Series returns **0.0** (identity element for addition).
- `mean()` of an all-NaN Series returns **NaN** because the average is undefined.

---

# Dropping a data

we can drop the duplicates but also we can drop a certain column or rows based on the conditions

1. drop columns 
2. drop rows 

here the `axis` is very important

## Axis

axis = 0 → move DOWN (↓)
axis = 1 → move ACROSS (→)

axis = 0 acts on the rows, so this means we get the result for each column. Hence axis = 0 movement along rows and column result 
axis = 1 acts on the columns, so this means we get the result for each row. Hence axis = 1 column movement along columns and row result

when we want to delete a column we are not moving up to down rather we remove from the set of values in the row and vice versa for removing rows. 

When we are doing operations on rows to get the result of a column then we move top to bottom 

| Axis | Movement | What is involved | Calculation result | Deletion effect |
|------|--------|-----------------|------------------|----------------|
| **axis = 0** | ↓ Down | Rows | Column-wise result | Removes **rows** |
| **axis = 1** | → Across | Columns | Row-wise result | Removes **columns** |

|----------|--------|

`right` keeps data from the right side merge


````python
pd.merge(temp,df_price, on=['ID','Order'],how='right')
````

````python
pd.merge(df_price,temp, on=['ID','Order'], how='right')
````


We can note the differnce between `outer` and `left` or `right`

`outer` make sure both the table contributes to the matching so id matched tablets got pushed up as ID matched

`left` or `right` doesn't do that, merge preserves left/right dataframe order

#  Function vs Output Type


##  1. Selection / Access

| Function | Operates On | Returns | Description |
|--------|------------|--------|-------------|
| `df['col']` | DataFrame | Series | Select one column |
| `df[['col1','col2']]` | DataFrame | DataFrame | Select multiple columns |
| `df.loc[]` | DataFrame | Series/DataFrame | Label-based selection |
| `df.iloc[]` | DataFrame | Series/DataFrame | Position-based selection |

## 2. Basic Info & Structure

| Function | Returns | Description |
|--------|--------|-------------|
| `df.head()` | DataFrame | First 5 rows |
| `df.tail()` | DataFrame | Last 5 rows |
| `df.info()` | None | Summary (prints only) |
| `df.describe()` | DataFrame | Stats summary |
| `df.shape` | tuple | (rows, cols) |
| `df.columns` | Index | Column names |
| `df.index` | Index | Row index |

## 3. Aggregations

| Function | Returns | Notes |
|--------|--------|------|
| `df.sum()` | Series | Column-wise sum |
| `df.mean()` | Series | Column-wise mean |
| `df.count()` | Series | Non-null count |
| `df.min()` / `df.max()` | Series | Min / Max |
| `df.agg()` | Series/DataFrame | Depends on input |

### `.agg()`

| Columns | Functions | Output |
|--------|----------|--------|
| 1 | 1 | scalar |
| 1 | many | Series |
| many | 1 | Series |
| many | many | DataFrame |

## 4. Column-Level (Series operations via DF)

| Function | Returns | Notes |
|--------|--------|------|
| `df['col'].sum()` | scalar | Single value |
| `df['col'].unique()` | ndarray | Unique values |
| `df['col'].nunique()` | scalar | Count |
| `df['col'].value_counts()` | Series | Frequency |

## 5. Missing Values

| Function | Returns |
|--------|--------|
| `df.dropna()` | DataFrame |
| `df.fillna()` | DataFrame |
| `df.isna()` | DataFrame (bool) |
| `df.notna()` | DataFrame (bool) |

##  6. Grouping

| Function | Returns |
|--------|--------|
| `df.groupby()` | GroupBy object |
| `grouped.sum()` | DataFrame/Series |
| `grouped['col'].sum()` | Series |
| `grouped.agg()` | DataFrame/Series |

`.sum()` is just one method, all the other methods like `.count()`,`.mean()`, etc. behave similarly  

| Operation | Column Selection | Functions | Output |
|----------|----------------|----------|--------|

`grouped = df.groupby('column_name')`

## Master map

### Step 1: Selecting?
- `df['col']` → Series  
- `df[['col1','col2']]` → DataFrame  

### Step 2: Aggregating?
- Yes:
  - 1 value → scalar  
  - Multiple columns → Series  
  - Multiple functions → DataFrame  

### Step 3: Groupby?
- Yes:
  - 1 value per group → Series  
  - Multiple values per group → DataFrame
