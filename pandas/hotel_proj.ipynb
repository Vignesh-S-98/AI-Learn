# Filling missing values 
We can fill the missing values with the follow procedures 
- For numerical column 
    1. Mean 
    2. Median
    3. Forward fill -> impute the missing value with the value that is before this entry i.e. a row above
    4. Backward fill -> impute the missing value with the value that is after this entry i.e. a row below

- For categorical column
    1. mode 
    2. assign a new name for the elements with `NaN` 

```python
df.isnull().any(axis=0).sum()
```

```python
def null_status():
    return df.isnull().sum().loc[lambda x: x > 0].sort_values(ascending=False)
```

```python
(df.isnull().sum()/df.shape[0]*100).loc[lambda x : x>0].sort_values(ascending=False)
```

Since company is having 94% data as a nan we dont want that column 

so we are droping that 

```python
df.drop("company",axis=1,inplace=True)
df.columns
```

```python
# df.isnull().sum()[df.isnull().sum()/df.shape[0]*100 > 0]
null_status()
```

children has very few missing value

lets explore that 

```python
# unique values in children column 
df['children'].value_counts()
```

Obs:- more number of people comes with 0 children so we are filling those missing values with 0 

```python
df.children  = df.children.fillna(0)
null_status()
```
